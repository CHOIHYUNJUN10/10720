```python
num=int(input("숫자를 입력하세요:"))
if 0==num%2:
  print("짝수입니다")
else:
  print("홀수입니다")
```

```python
year=int(input("년도를 입력하세요:"))
if (year%4==0 and year%100!=0) or (year%400==0):
  print("윤년")
else:
  print("윤년X")
```

```python
age=int(input("나이를 입력하세요:"))
if age >=18:
  print("성인 교통카드가 발급됩니다.")
  print("일반 교통 요금이 적용됩니다.")
else:
  print("어린이/청소년 교통카드가 발급됩니다.")
  print("교통 요금이 할인됩니다.")
```

```python
month=int(input("몇월인지 입력하세요:"))
if month==12  or month <=2 and month >0:
  print("겨울")
elif month >= 6 and month <= 8:
  print("여름")
elif  month>=3 and month <=5:
  print("봄")
elif month <=0 or month>=13:
  print("잘못된 입력입니다.1~12사이로 입력 해주세요")
else:
  print("가을")
```

```python
점수 = int(input("시험점수를 입력하세요"))
if 점수>100 or 점수<0:
  print("제대로 된 점수를 입력하세요.")
elif 점수<=100 and 점수 >= 90:
  print("A")
elif 점수<90 and 점수 >= 80:
  print("B")
elif 점수<80 and 점수 >= 70:
  print("C")
elif 점수<70 and 점수 >= 60:
  print("D")
else:
  print("F")
```
