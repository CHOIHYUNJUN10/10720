```python
a=int(input("숫자를 입력하세요:"))
b=int(input("숫자를 입력하세요:"))
c=input("사칙연산을 입력하세요")
def cal(a,op,b):
  if op=='+':
    return a+b
  elif op=='-':
    return a-b
  elif op=='*':
    return a*b
  else:
    if b==0:
      print("0으로는 나눌수없습니다")
    else:
      return a/b
print(cal(a,c,b))
```
